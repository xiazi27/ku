```python
import pandas as pd
import numpy as np
```

```python
df = pd.read_csv(r'C:\Users\tjx\Downloads\titanic.csv')
df.head()
```

```python
print("原始列名:", df.columns)
```

```python
print("每列缺失值的数量")
print(df.isnull().sum())
```

```python
missing_percent = (df.isnull().sum() / len(df)) * 100
```

```python
# 设定阈值，删除缺失超过30%的列
threshold = 30
columns_to_keep = missing_percent[missing_percent <= threshold].index
df = df[columns_to_keep]
```

```python
# 重新计算剩余列的缺失比例
missing_percent = (df.isnull().sum() / len(df)) * 100
```

```python
# 按照比例和类型填充缺失值
for col in df.columns:
    if df[col].isnull().sum() == 0:
        continue  # 没有缺失，跳过

    perc = missing_percent[col]

    if perc <= 5:
        # 少量缺失：使用众数
        fill_value = df[col].mode()[0]
    elif perc <= 30:
        # 中等缺失：根据类型选择填充值
        if df[col].dtype in ['float64', 'int64']:
            fill_value = df[col].median()
        else:
            fill_value = df[col].mode()[0]

    # 填充缺失值
    df[col] = df[col].fillna(fill_value)
```

```python
print("缺失值处理后的数据：")
print(df.head())
```

```python
print("Age列的描述统计：\n", df['Age'].describe())
print("\n是否包含小数：", df['Age'].apply(float.is_integer).value_counts() if df['Age'].notnull().all() else "含有缺失值")
```

```python
# 去掉不在 1~100 之间的值
df = df[(df['Age'] >= 1) & (df['Age'] <= 100)]

# 四舍五入并转为整数（或直接取整）
df['Age'] = df['Age'].round().astype(int)

# 查看结果
print("处理后 Age 列描述：\n", df['Age'].describe())
print("\n前几行 Age：\n", df['Age'].head())
```

```python
df['Sex'] = df['Sex'].str.lower()
df = df[df['Sex'].isin(['male', 'female'])]
df = df[df['Survived'].isin([0, 1])]
print("\n处理后的数据：")
print(df.head())
```

```python
df.boxplot(
    column='Age',
    by='Survived',
    grid=False,
    figsize=(8, 5),
    patch_artist=True,
    boxprops={'facecolor': 'lightblue'},
    medianprops={'color': 'red'}
)
```

```python
survival_by_class = df.groupby(['Pclass', 'Survived']).size().unstack()
survival_by_class.plot(
    kind='bar',
    title='Survival Count by Passenger Class',
    xlabel='Passenger Class',
    ylabel='Count',
    rot=0
)
```
